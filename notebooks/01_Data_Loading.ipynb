```python
import sqlite3
import pandas as pd
from pathlib import Path

# Explicit absolute path to ensure it lands inside your database folder
project_dir = Path.cwd()
if project_dir.name == "notebooks":
    project_dir = project_dir.parent
db_path = project_dir / "database" / "retail.db"
db_path.parent.mkdir(parents=True, exist_ok=True)

conn = sqlite3.connect(db_path)

# Automatically export all DataFrames currently in memory to retail.db
saved_count = 0
for name, obj in list(globals().items()):
    if isinstance(obj, pd.DataFrame) and not name.startswith('_'):
        table_name = name.replace("olist_", "").replace("_dataset", "")
        obj.to_sql(table_name, conn, if_exists="replace", index=False)
        print(f"✓ Saved table: '{table_name}' ({len(obj)} rows)")
        saved_count += 1

conn.close()
print(f"\nDone! Successfully created {db_path} with {saved_count} tables.")
```
